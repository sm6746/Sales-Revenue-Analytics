```python
import pandas as pd

# Reading Sales_Dataset.csv
Sales_Dataset = pd.read_csv('Sales_Dataset.csv')
print(f'✓ Loaded Sales_Dataset: {len(Sales_Dataset)} rows')
display(Sales_Dataset)
```

```python
import pandas as pd

# Make a copy of the dataframe to ensure the original is untouched
df_cleaned = Sales_Dataset.copy()

# Convert 'Order Date' to datetime format for time-based analysis
df_cleaned['Order Date'] = pd.to_datetime(df_cleaned['Order Date'])

# Check for and display the number of missing values in each column
print("Missing values in each column:")
print(df_cleaned.isnull().sum())

# Remove duplicate rows to ensure data accuracy
df_cleaned.drop_duplicates(inplace=True)
print(f"\nRemoved duplicates, new shape: {df_cleaned.shape}")

# Extract year and month from 'Order Date' for filtering and aggregation
df_cleaned['Year'] = df_cleaned['Order Date'].dt.year
df_cleaned['Month'] = df_cleaned['Order Date'].dt.month

print("\nData cleaning and preparation complete.")
display(df_cleaned)

```

```python
import pandas as pd
import altair as alt

# Calculate monthly profit and revenue
monthly_profit_margin = df_cleaned.groupby(['Year', 'Month']).agg(
    Total_Profit=('Profit', 'sum'),
    Total_Revenue=('Amount', 'sum')
).reset_index()

# Calculate the profit margin for each month
monthly_profit_margin['Profit_Margin'] = (monthly_profit_margin['Total_Profit'] / monthly_profit_margin['Total_Revenue']) * 100

# Create a 'Date' column for proper time-series plotting
monthly_profit_margin['Date'] = pd.to_datetime(monthly_profit_margin[['Year', 'Month']].assign(day=1))

# Create the line chart to show the trend of monthly profit margin
profit_margin_trend_chart = alt.Chart(monthly_profit_margin).mark_line(point=True).encode(
    x=alt.X('Date:T', title='Month'),
    y=alt.Y('Profit_Margin:Q', title='Profit Margin (%)'),
    tooltip=['Date:T', 'Profit_Margin:Q']
).properties(
    title='Monthly Profit Margin Trend'
)

display(profit_margin_trend_chart)
```

```python
import altair as alt
import pandas as pd

# Group data by year to analyze yearly sales performance
yearly_sales = df_cleaned.groupby('Year')['Amount'].sum().reset_index()

# Create a bar chart to visualize yearly sales performance
chart = alt.Chart(yearly_sales).mark_bar().encode(
    x=alt.X('Year:O', title='Year'),
    y=alt.Y('Amount:Q', title='Total Sales'),
    tooltip=['Year', 'Amount']
).properties(
    title='Yearly Sales Performance'
)

display(chart)
```

```python
import altair as alt

# Group data by state to analyze region-wise performance
region_performance = df_cleaned.groupby('State')['Amount'].sum().reset_index()

# Create a pie chart to visualize region-wise performance
chart = alt.Chart(region_performance).mark_arc().encode(
    theta=alt.Theta(field="Amount", type="quantitative"),
    color=alt.Color(field="State", type="nominal"),
    tooltip=['State', 'Amount']
).properties(
    title='Region-wise Performance'
)

display(chart)
```

```python
import altair as alt
import pandas as pd

# Group data by category to analyze category-wise distribution
category_distribution = df_cleaned['Category'].value_counts().reset_index()
category_distribution.columns = ['Category', 'Count']

# Create a bar chart to visualize the category distribution
chart = alt.Chart(category_distribution).mark_bar().encode(
    x=alt.X('Category:N', title='Category'),
    y=alt.Y('Count:Q', title='Count'),
    tooltip=['Category', 'Count']
).properties(
    title='Distribution of Categories'
)

display(chart)
```

```python
import altair as alt
import pandas as pd

# Group data by payment mode to analyze its distribution
payment_mode_distribution = df_cleaned['PaymentMode'].value_counts().reset_index()
payment_mode_distribution.columns = ['PaymentMode', 'Count']

# Create a pie chart to visualize the payment mode distribution
chart = alt.Chart(payment_mode_distribution).mark_arc().encode(
    theta=alt.Theta(field="Count", type="quantitative"),
    color=alt.Color(field="PaymentMode", type="nominal"),
    tooltip=['PaymentMode', 'Count']
).properties(
    title='Distribution of Payment Modes'
)

display(chart)
```

```python
import altair as alt
import pandas as pd

# Find the most purchased item for each customer
most_purchased_item = df_cleaned.groupby(['CustomerName', 'Sub-Category']).size().reset_index(name='counts')
idx = most_purchased_item.groupby(['CustomerName'])['counts'].idxmax()
top_customers_df = most_purchased_item.loc[idx].nlargest(10, 'counts')


# Create the chart
chart = alt.Chart(top_customers_df).mark_bar().encode(
    x=alt.X('CustomerName:N', sort='-y', title='Customer Name'),
    y=alt.Y('counts:Q', title='Number of Purchases'),
    color=alt.Color('Sub-Category:N', title='Most Purchased Item'),
    tooltip=['CustomerName', 'Sub-Category', 'counts']
).properties(
    title='Top 10 Customers and Their Most Purchased Item'
)

display(chart)
```
